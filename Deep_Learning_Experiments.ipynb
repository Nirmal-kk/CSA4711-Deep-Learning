### 1. Image Classification: Fully Connected Layer Operations


```
import numpy as np
np.random.seed(42)
image_matrix = np.random.randint(0, 256, size=(28, 28))
print(f"Original Image Shape: {image_matrix.shape}")
flattened_vector = image_matrix.flatten()
print(f"Flattened Vector Shape: {flattened_vector.shape}")
weights = np.random.randn(784, 10) * 0.01
bias = np.zeros(10)
output = np.dot(flattened_vector, weights) + bias
print(f"Layer Output Shape: {output.shape}")
print("Calculated Output Projections (logits):\n", output)
```

### 2. Student Performance Prediction


```
attendance = np.random.uniform(75, 100, (100, 1))
assignments = np.random.uniform(60, 100, (100, 1))
internals = np.random.uniform(20, 50, (100, 1))

X = np.hstack((attendance, assignments, internals))
print(f"Feature Matrix X Shape: {X.shape} (100 Students, 3 Features)")
W = np.array([[0.3], [0.5], [0.8]])
b = 5.0
predictions = np.dot(X, W) + b
print(f"Predictions Shape: {predictions.shape}")
print("First 5 Predicted Student Scores:\n", predictions[:5])
```

### 3. E-commerce Recommendation System


```
product_features = np.random.randn(500, 20)
norms = np.linalg.norm(product_features, axis=1, keepdims=True)
normalized_features = product_features / norms
similarity_matrix = np.dot(normalized_features, normalized_features.T)
target_product_id = 0
similar_products = np.argsort(-similarity_matrix[target_product_id])
print(f"Similarity Matrix Shape: {similarity_matrix.shape}")
print(f"Top 5 products recommended for Product {target_product_id}: {similar_products[1:6]}")
```

### 4. Medical Image Diagnosis


```
batch_features = np.random.randn(10, 50)
feature_mask = np.random.uniform(0.8, 1.2, size=(1, 50))
masked_features = batch_features * feature_mask
diagnosis_weights = np.random.randn(50, 2)
diagnosis_bias = np.array([0.1, -0.1])
logits = np.dot(masked_features, diagnosis_weights) + diagnosis_bias
print(f"Logits shape for diagnosis output: {logits.shape}")
print("Sample diagnostic output scores (First 3 cases):\n", logits[:3])
```
